```python
notas = ['nome1', 9.0, 7.0, 10.0,'nome2', 3.0, 5.0, 9.0,'nome3', 7.0, 1.0, 10.0,]

nomes = []
notas_juntas = []

for i in range(len(notas)):
    if i % 4  == 0:
        nomes.append(notas[i])
    else:
        notas_juntas.append(notas[i])
```
