```python
# Funções Built-infunction

nota = {'1 Trimestre': 8.6 ,'2 Trimestre': 9.2,'3 Trimestre': 7.7}

display(nota)
```

```python
# calculando a soma 

soma = 0

for notas in nota.values():
    soma += notas

soma

```

```python
# somando com built-funcions

somatorio = sum(nota.values())
somatorio
```

```python
# tamanho com built-funcions

contagem = len(nota)
contagem

```

```python
# calculando média de notas

media_notas = somatorio / contagem
media_notas = round(media_notas,1)

print(media_notas)
```
